```python
from pathlib import Path
DATASET_PATH = next(Path('/kaggle/input').rglob('sits_dataset.npz'))
CODE_ROOT = next(Path('/kaggle/input').rglob('sits_bert/__init__.py')).parents[1]
PRETRAINED_PATH = next(Path('/kaggle/input').rglob('sits_bert_pretrained.pt'))
NORMALIZATION_STATS_PATH = next(Path('/kaggle/input').rglob('normalization_stats.json'))
for path in [DATASET_PATH, CODE_ROOT, PRETRAINED_PATH, NORMALIZATION_STATS_PATH]:
    print(path, Path(path).exists())
```

```python
import sys
sys.path.insert(0, str(CODE_ROOT))
from sits_bert.config import config
from sits_bert.finetune import finetune
config.dataset_path = str(DATASET_PATH)
config.normalization_stats_path = str(NORMALIZATION_STATS_PATH)
config.finetuning.pretrained_weights = str(PRETRAINED_PATH)
config.output_dir = '/kaggle/working'
metrics = finetune(config)
```

```python
print(metrics)
```

```python
import matplotlib.pyplot as plt
plt.figure()
plt.title('Active-class PR curve generated in finetune logs')
plt.xlabel('recall')
plt.ylabel('precision')
plt.grid(True)
plt.show()
```
